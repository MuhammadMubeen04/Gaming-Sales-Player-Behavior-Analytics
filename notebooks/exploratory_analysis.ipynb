# Gaming Sales & Player Behavior – Exploratory Analysis

Quick exploratory data analysis notebook for the Gaming Analytics project.

**Data:** 1,200 games (2012–2025) with sales, ratings, genre, platform, and playtime.

## 1. Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams["figure.figsize"] = (11, 6)

print("Libraries loaded.")
```

## 2. Load Data

```python
df = pd.read_csv("../data/raw/games_data.csv")
print(f"Shape: {df.shape}")
df.head()
```

## 3. Basic Info

```python
df.info()
```

```python
df.describe()
```

```python
print("Missing values:")
print(df.isnull().sum())
```

## 4. Categorical Overview

```python
print("Platform counts:")
print(df["Platform"].value_counts())
print("\nGenre counts:")
print(df["Genre"].value_counts())
```

## 5. Sales by Platform

```python
platform_sales = df.groupby("Platform")["Global_Sales"].sum().sort_values(ascending=True)
platform_sales.plot(kind="barh", color="#4e79a7")
plt.title("Total Global Sales by Platform (Million Units)")
plt.xlabel("Sales (Millions)")
plt.tight_layout()
plt.show()
```

## 6. Sales by Genre

```python
genre_sales = df.groupby("Genre")["Global_Sales"].sum().sort_values(ascending=True)
genre_sales.plot(kind="barh", color="#59a14f")
plt.title("Total Global Sales by Genre (Million Units)")
plt.xlabel("Sales (Millions)")
plt.tight_layout()
plt.show()
```

## 7. Yearly Trend

```python
yearly = df.groupby("Year").agg(Sales=("Global_Sales", "sum"), Games=("GameID", "count"))
fig, ax1 = plt.subplots(figsize=(12, 5))
ax1.bar(yearly.index, yearly["Sales"], color="#4e79a7", alpha=0.8)
ax1.set_ylabel("Sales (Millions)", color="#4e79a7")
ax2 = ax1.twinx()
ax2.plot(yearly.index, yearly["Games"], color="#e15759", marker="o")
ax2.set_ylabel("Games Released", color="#e15759")
ax1.set_title("Yearly Sales & Number of Games Released")
plt.tight_layout()
plt.show()
```

## 8. Critic vs User Scores

```python
sns.scatterplot(data=df, x="Critic_Score", y="User_Score", hue="Genre", alpha=0.6)
plt.title("Critic Score vs User Score by Genre")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left", fontsize=8)
plt.tight_layout()
plt.show()
```

## 9. Top 10 Games by Sales

```python
top10 = df.nlargest(10, "Global_Sales")[["Name", "Platform", "Genre", "Year", "Global_Sales", "Critic_Score"]]
top10
```

## 10. Quick Insights

```python
print("Total games:", len(df))
print("Total sales (millions):", round(df["Global_Sales"].sum(), 1))
print("Avg critic score:", round(df["Critic_Score"].mean(), 1))
print("Avg user score:", round(df["User_Score"].mean(), 2))
print("\nTop platform by sales:", df.groupby("Platform")["Global_Sales"].sum().idxmax())
print("Top genre by sales:", df.groupby("Genre")["Global_Sales"].sum().idxmax())
```
